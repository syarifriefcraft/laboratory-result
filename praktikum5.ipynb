```python
def buat_biodata(nama, umur, jurusan):
  # menampilkan ketiga data
  print(f"nama: {nama}")
  print(f"nama: {umur}")
  print(f"jurusan: {jurusan}")

# memanggil sekali dengan positional argument
buat_biodata("syarif", 18,"SISTEM INFORMASI")

print("-" * 18)   #pembatasan tampilan

#memanggil sekali dengan keyword argument dengan urutan berbeda
buat_biodata(jurusan="sistem informasi", nama="syarif",umur=18)
```

```python
biaya_admin = 2500   # variabel Global
def hitung_total(harga):
  total = harga + biaya_admin  #variabel lokal 'total
  return total
# memanggil fungsi untuk harga 15000
hasil_total = hitung_total(15000)
print("total", hasil_total)
```

```python
def konverensi_menit_ke_detik(menit):
  """
  mengonversi menit menjadi detik.
  parameter:
      menit(int/float):jumlah waktu dalam satuan menit yang akan di konversi.

      nilai kembalian:
      int/float:jumlah waktu hasil konversi dalam satuan detik.
      """
  return menit * 60
print(konversi_menit_ke_detik(3))
print(konversi_menit_ke_detik.__doc__)
```

```python
def hitung_subtotal(harga, jumlah):
    return harga * jumlah

def hitung_diskon(subtotal):
    if subtotal >= 100000:
        return subtotal * 0.10
    else:
        return 0

def tampilkan_struk(nama_barang, subtotal, diskon):
    total_akhir = subtotal - diskon
    print("=== STRUK PEMBELIAN ===")
    print(f"Nama Barang : {nama_barang}")
    print(f"Subtotal    : Rp {subtotal:,.0f}")
    print(f"Diskon      : Rp {diskon:,.0f}")
    print(f"Total Akhir : Rp {total_akhir:,.0f}")
    print("=======================")

# Alur Pengujian Program
nama_barang = "Buku Gambar"
harga_barang = 25000
jumlah_beli = 5

# 1. Hitung Subtotal
subtotal = hitung_subtotal(harga_barang, jumlah_beli)

# 2. Hitung Diskon berdasarkan Subtotal
diskon = hitung_diskon(subtotal)

# 3. Tampilkan Struk Akhir
tampilkan_struk(nama_barang, subtotal, diskon)
```
