praktikum01_for_range 01

```python
# Menampilkan bilangan 1 sampai 20
print("Bilangan 1 sampai 20:")
for i in range(1, 21):
    print(i, end=" ")
print("\n")

# Menampilkan kelipatan 3 dari 3 sampai 30
print("Kelipatan 3 dari 3 sampai 30:")
for i in range(3, 31, 3):
    print(i, end=" ")
print()
```

praktikum02_string 02

```python
kalimat = input("Masukkan sebuah kalimat: ")
hitung_vokal = 0

for huruf in kalimat.lower():
    if huruf in "aiueo":
        hitung_vokal += 1

print("Jumlah huruf vokal:", hitung_vokal)
```

praktikum03_validasi 03

```python
# Ganti token sesuai dengan kode kelompok kamu
kode_token = ""
token_rahasia = "VVIP2026"

while kode_token != token_rahasia:
    kode_token = input("Masukkan kode token akses ujian: ")
    if kode_token != token_rahasia:
        print("Kode token salah, silakan coba lagi.")

print("Token valid. Anda dapat memulai ujian.")
```

praktikum04_break_continue 04

```python
# Latihan continue (menampilkan angka ganjil 1-10)
print("Angka ganjil 1-10:")
for angka in range(1, 11):
    if angka % 2 == 0:
        continue
    print(angka, end=" ")
print("\n")

# Modifikasi break: mencari bilangan 1-200 yang habis dibagi 4 dan 6
for angka in range(1, 201):
    if angka % 4 == 0 and angka % 6 == 0:
        print(f"Bilangan pertama yang habis dibagi 4 dan 6 adalah: {angka}")
        break
```

praktikum05_pola 05

```python
print("--- Pola Segitiga ---")
for baris in range(1, 6):
    for kolom in range(baris):
        print("*", end="")
    print()

print("\n--- Pola Persegi ---")
for baris in range(1, 5):
    for kolom in range(1, 6):
        print("*", end="")
    print()

print("\n--- Pola Segitiga Terbalik (Pengembangan) ---")
for baris in range(5, 0, -1):
    for kolom in range(baris):
        print("*", end="")
    print()
```

praktikum06_for_else 06

```python
bilangan = int(input("Masukkan sebuah bilangan bulat: "))

if bilangan > 1:
    for i in range(2, bilangan):
        if bilangan % i == 0:
            print(f"{bilangan} bukan bilangan prima")
            break
    else:
        print(f"{bilangan} adalah bilangan prima")
else:
    print(f"{bilangan} bukan bilangan prima")
```
