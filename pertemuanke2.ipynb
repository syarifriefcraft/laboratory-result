praktikum 01

```python
# Program Menghitung Luas dan Keliling Persegi Panjang

panjang = float(input("Masukkan panjang: "))
lebar = float(input("Masukkan lebar: "))

luas = panjang * lebar
keliling = 2 * (panjang + lebar)

print("Luas =", luas)
print("Keliling =", keliling)
```

praktikum 02

```python
# Program Konversi Suhu Fahrenheit ke Celsius

fahrenheit = float(input("Masukkan suhu Fahrenheit: "))

celsius = (fahrenheit - 32) * 5 / 9

print(f"{fahrenheit} derajat Fahrenheit setara dengan {celsius:.2f} derajat Celsius")
```

praktikum 03

```python
# Program Total Belanja dengan Diskon

harga_satuan = float(input("Masukkan harga satuan: "))
jumlah = int(input("Masukkan jumlah barang: "))
persen_diskon = float(input("Masukkan persentase diskon (%): "))

subtotal = harga_satuan * jumlah
nominal_diskon = subtotal * (persen_diskon / 100)
total_bayar = subtotal - nominal_diskon

print(f"Subtotal: Rp {subtotal:,.2f}")
print(f"Diskon: Rp {nominal_diskon:,.2f}")
print(f"Total Bayar: Rp {total_bayar:,.2f}")
```

```python
# Program Konversi Detik ke Jam, Menit, Detik

total_detik = int(input("Masukkan total detik: "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print(f"{jam:02d}:{menit:02d}:{detik:02d}")
```
